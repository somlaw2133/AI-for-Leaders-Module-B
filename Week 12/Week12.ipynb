# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

The two approaches make different types of comparisons easier.
Using two separate graphs makes each week easy to examine on its own because the lines do not overlap. This can make the individual pattern for each week clearer. However, it is harder to directly compare the two weeks because the viewer has to move back and forth between the graphs.
Putting both weeks on the same graph makes it easier to compare visitor numbers on the same days and quickly see where one week was higher or lower than the other. The disadvantage is that the graph can become more cluttered, especially if there are many lines or if the values overlap.
For this example, I would prefer the combined graph because there are only two weeks, so the comparison is still easy to follow and the differences between the weeks are more obvious.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

years = np.arange(2008, 2015)

# Approximate values based on Figure 8.6
product_a = [370, 360, 345, 325, 305, 280, 260]
product_b = [360, 350, 330, 310, 290, 270, 250]
product_c = [np.nan, np.nan, 180, 210, 205, 195, 180]
product_d = [np.nan, np.nan, np.nan, 200, 230, 220, 210]
product_e = [np.nan, np.nan, np.nan, np.nan, np.nan, 170, 215]

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(years, product_a, linewidth=2)
ax.plot(years, product_b, linewidth=2)
ax.plot(years, product_c, linewidth=2)
ax.plot(years, product_d, linewidth=2)
ax.plot(years, product_e, linewidth=2)

# Direct labels at the end of each line
ax.text(2014.1, product_a[-1], "A", va="center")
ax.text(2014.1, product_b[-1], "B", va="center")
ax.text(2014.1, product_c[-1] - 5, "C", va="center")
ax.text(2014.1, product_d[-1] + 7, "D", va="center")
ax.text(2014.1, product_e[-1] - 5, "E", va="center")

ax.set_title(
    "Average Retail Product Price per Year",
    loc="left",
    fontsize=14
)

ax.set_xlim(2008, 2014.5)
ax.set_ylim(0, 500)

ax.set_xticks(years)
ax.set_yticks(np.arange(0, 501, 100))
ax.set_yticklabels([f"${x}" for x in np.arange(0, 501, 100)])

# Remove clutter
ax.grid(False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

I chose to recreate Figure 8.6, “Eliminate clutter,” from page 194 of *Storytelling With Data*. The graph shows how the average retail prices of five products changed over time. I chose this figure because it shows how a graph can become easier to understand by removing unnecessary visual elements. The chart does not use gridlines or a separate legend, and each line is labeled directly. This makes it easier to follow the different products without having to look back and forth between the graph and a legend.
I think this example does a good job of showing that simplifying a graph does not mean removing useful information. Instead, it removes distractions so the audience can focus more on the actual data.
